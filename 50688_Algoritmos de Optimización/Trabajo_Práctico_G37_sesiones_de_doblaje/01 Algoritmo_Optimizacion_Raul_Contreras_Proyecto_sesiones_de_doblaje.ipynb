Descripción del problema:

**Organizar sesiones de doblaje**
• Se precisa coordinar el doblaje de una película. Los actores del doblaje deben coincidir en las tomas en las que sus personajes aparecen juntos en las diferentes tomas. Los actores de doblaje cobran todos la misma cantidad por cada día que deben desplazarse hasta el estudio de grabación independientemente del número de tomas que se graben. No es posible grabar más
de 6 tomas por día. El objetivo es planificar las sesiones por día de manera que el gasto por los servicios de los actores de doblaje sea el menor posible. Los datos son:

Número de actores: 10
Número de tomas : 30
Actores/Tomas : https://bit.ly/36D8IuK
- 1 indica que el actor participa en la toma
- 0 en caso contrario


....

### 1Análisis de Complejidad
- Complejidad temporal: Está determinada por el número de iteraciones ($I$) destinadas a mejorar la agenda, la cantidad de tomas ($T$) y el total de actores ($A$).
- Cálculo de coste: En cada evaluación de una agenda, el algoritmo recorre la totalidad de las tomas y actores presentes en la matriz.Bucle principal: Dado que este cálculo de coste se ejecuta en cada una de las 1,000 iteraciones, el tiempo global resulta de multiplicar las iteraciones por el tamaño de la matriz.
- Conclusión: Se trata de un algoritmo rápido y eficiente para el volumen de datos manejado en este proyecto de doblaje.



### 2 Justificación de la Estructura de Datos
Se ha seleccionado una **lista de listas** anidada (`schedule_days`), donde cada sublista representa una jornada de grabación que contiene los identificadores numéricos de las tomas asignadas por dias que en este caso son 6.

Esta estructura resulta idónea por las siguientes razones:
1. **Modelado directo:** Refleja de forma natural la restricción de capacidad máxima por día. Que para este trabajo son hasta 6 tomas.
2. **Eficiencia en operadores de vecindario:** Facilita el intercambio de elementos entre diferentes días (operación *swap*) de manera directa mediante índices, lo cual es fundamental para los algoritmos de búsqueda local y recocido simulado.
3. **Compatibilidad:**Permite consultar directamente el DataFrame mediante los índices de las tomas para evaluar las apariciones de los actores y calcular el coste al instante.

````python
import os
import subprocess

def mi_push_global(ruta_drive, nombre_repo, mensaje_commit):
    raiz_git = os.path.join(ruta_drive, nombre_repo)

    if not os.path.exists(raiz_git):
        print(f"Error: No se encuentra la carpeta en la ruta: {raiz_git}")
        return

    # 1. Configurar Git
    subprocess.run(['git', '-C', raiz_git, 'config', 'core.quotepath', 'false'])

    # 2. Ver estado antes de tocar nada
    resultado_estado = subprocess.run(
        ['git', '-C', raiz_git, 'status', '--porcelain', '--untracked-files=all'],
        capture_output=True,
        text=True,
        encoding='utf-8'
    )

    if not resultado_estado.stdout.strip():
        print("--- Estado actual ---")
        print("Todo al día en todo el repositorio, no hay nada que subir. 🚀")
        return

    print(f"--- Archivos pendientes en [{nombre_repo}] ---")
    print(resultado_estado.stdout)

    respuesta = input("¿Quieres hacer el commit y push de TODOS estos cambios a GitHub? (s/n): ").strip().lower()

    if respuesta == 's':
        # 3. Secuencia estricta: Add -> Commit -> Push
        subprocess.run(['git', '-C', raiz_git, 'add', '.'])

        # Intentar commit (si da error porque no hay cambios nuevos, lo manejamos)
        commit_res = subprocess.run(['git', '-C', raiz_git, 'commit', '-m', mensaje_commit], capture_output=True, text=True)

        # Hacer el push definitivo
        resultado_push = subprocess.run(
            ['git', '-C', raiz_git, 'push'],
            capture_output=True,
            text=True,
            encoding='utf-8'
        )

        if resultado_push.returncode == 0:
            print(f"¡El repositorio '{nombre_repo}' se sincronizó con éxito en GitHub! ✨")
        else:
            print("⚠️ Hubo un detalle al subir:")
            print(resultado_push.stderr)
    else:
        print("Operación cancelada.")

# Ejecutar
mi_push_global(MI_RUTA_DRIVE, MI_REPOSITORIO, "Actualización general desde Colab")
````

# **Paso 1: Lectura de daros**

 - La Matriz: Se lee la matriz Tomas_x_Actores: 10 actores y 30 tomas disponibles. Los datos estan representados por 0s y 1s, donde 1 representa que el actor debe asistir a esa toma y cero representa que el actor no se necesita.
 - La Fuente de Datos: Es un archivo externo tipo .xls y se crea un  DataFrame (df) de Pandas.



````python
# Librerias Necesarias para el projecto
import pandas as pd
import numpy as np
import os
import random
import math
import copy
````

````python
# --- Paso 1: Carga y Limpieza de Datos ---
def carga_datos_archivo():
    """
    Carga el archivo Excel de la práctica, realiza la limpieza de celdas,
    convierte las tomas en códigos estructurados, lo procesa internamente
    como df_file y lo retorna para asignarlo externamente.
    """
    # --- Entorno de ejecución y carga de datos ---
    IN_COLAB = 'google.colab' in str(get_ipython()) if 'get_ipython' in globals() else False
    archivo_datos = 'Datos problema doblaje(30 tomas, 10 actores).xlsx'

    try:
        if IN_COLAB:
            print("Ejecutando en Google Colab...")
            from google.colab import drive
            drive.mount('/content/drive')
            ruta = '/content/drive/MyDrive/Colab Notebooks/02 - 03MIAR_04_B_2026-27_Algoritmos de Optimización/01 VIU 2026 AG_Algoritmos de Optimizacion /PROYECTO (Trabajo Práctico) 1a Conv_V2/'
            full_ruta = os.path.join(ruta, archivo_datos)
        else:
            print("Ejecutando en entorno local...")
            ruta = './'
            full_ruta = os.path.join(ruta, archivo_datos)

        if not os.path.exists(full_ruta):
            raise FileNotFoundError(f"El archivo '{archivo_datos}' no existe en la ruta: {full_ruta}")

        # Lectura inicial del archivo de Excel, filtrando a 30 tomas
        df = pd.read_excel(full_ruta, header=1).head(30)

        # Asegura valores binarios (0 y 1) estrictos en las columnas de actores
        # Explicación del dataframe:
        # - df.iloc[:, 1:]: Selecciona todas las filas (:) y desde la columna 1 en adelante (1:),
        #   dejando intacta la primera columna (tomas) y tomando solo las columnas de los actores.
        # - .map(lambda x: 1 if x == 1 else 0): Recorre cada celda; si el valor es 1, lo deja en 1;
        #   de lo contrario (un 22, un texto, un NaN o vacío), lo convierte en 0 de manera segura.
        df.iloc[:, 1:] = df.iloc[:, 1:].map(lambda x: 1 if x == 1 else 0)

        # Conversión de seguridad para garantizar enteros limpios en todo el DataFrame
        df = df.apply(pd.to_numeric, errors='coerce').fillna(0).astype(int)

        # Asignación correcta de nombres de columnas estándar
        df.columns = ['Toma_#'] + [f"Actor_{i}" for i in range(1, len(df.columns))]

        # Transformación clave: Convertimos la columna de tomas en códigos reales (ej. 'Toma_01', 'Toma_02'...)
        # Esto le da robustez profesional al modelo para que no dependa solo de índices numéricos frágiles.
        df['Toma_#'] = [f"Toma_{i:02d}" for i in range(1, len(df) + 1)]
        #for i in range(1, len(df) + 1): Crea números del 1 al 30).
        #f"Toma_{i:02d}":  f-string arma un texto combinando la palabra "Toma_" con el número.
        #El detalle clave es el :02d: significa que le obliga a ocupar 2 #dígitos con un cero a la izquierda si es menor de 10. Por ejemplo: #Toma_01, Toma_02, ..., Toma_10.
        #'Toma_01' a la 'Toma_30') y reemplaza por completo la columna de identificadores de tu tabla.

        # Filtrado de columnas necesarias almacenándolo internamente como df_file (10 Actores + 1 columna de toma)
        num_col = 11
        if df.shape[1] > num_col:
            df_file = df.iloc[:, :num_col].copy()
        else:
            df_file = df.copy()



        return df_file

    except Exception as error:
        print(f"Se ha producido un error durante la carga: {error}")
        return None

# Ejecución --->
df_clean = carga_datos_archivo()
print("\n Verificación de Datos Simulados --->")
print(f"Dimensiones correctas de la matriz (Filas, Columnas): {df_clean.shape}")
print(f"Nombres de las columnas: {df_clean.columns.tolist()}")
print()
print(df_clean.head(2))
````

# **Paso 2: Generar Datos de Prueba**
  Se implementa una rutina que genera datos aleatorios para realizar las pruebas

````python
# --- Generación de datos de prueba simulados ---
def generar_datos_prueba(num_takes=30, num_actors=10, seed=42):
    """
    Genera un DataFrame sintáctico mediante distribución binomial para simular
    la matriz de participación de actores en las tomas, permitiendo validar
    el comportamiento del modelo ante diferentes escenarios.
    """
    # 1. Configurar la semilla de aleatoriedad para reproducibilidad
    np.random.seed(seed)

    # 2. Generar matriz binaria de participación (0s y 1s) con una probabilidad del 40%
    mock_matrix = np.random.binomial(1, 0.4, size=(num_takes, num_actors))

    # 3. Crear la columna identificadora para las tomas
    toma_col = np.arange(1, num_takes + 1).reshape(-1, 1)

    # 4. Unir horizontalmente la columna de tomas con la matriz de actores
    full_data = np.hstack((toma_col, mock_matrix))

    # 5. Definir la nomenclatura de las columnas
    column_names = ['Toma_#'] + [f"Actor_{i}" for i in range(1, num_actors + 1)]

    # 6. Construir el DataFrame definitivo
    df_mock = pd.DataFrame(full_data, columns=column_names)
    df_mock['Toma_#'] = [f"Toma_{i:02d}" for i in range(1, len(df_mock) + 1)]
    #print(f"Dimensiones correctas de la matriz Mock (Filas, Columnas): {df_mock.shape}")
    return df_mock



# --- Ejecución y verificación ---
df_clean = generar_datos_prueba(num_takes=33, num_actors=15, seed=42)
print("\n Verificación de Datos Simulados --->")
print(f"Dimensiones correctas de la matriz (Filas, Columnas): {df_clean.shape}")
print(f"Nombres de las columnas: {df_clean.columns.tolist()}")
print(df_clean.head(2))
````

# **Paso 4: El Creamos la agenda y la imprimimos**
   - Crear Agenda
   - Imprimir Agenda

````python
# --- Paso 1: Función para Crear la Agenda (Inicial o la que sea) ---
def crear_agenda_doblaje(matriz_actores, max_tomas_por_dia=6):
    """
    Agrupa las tomas disponibles en jornadas diarias utilizando los códigos
    reales de la columna 'Toma_#', respetando el límite máximo establecido por día.
    """
    agenda_dias = []
    dia_actual = []

    # Extraemos directamente los códigos de texto de la columna 'Toma_#'
    codigos_tomas = matriz_actores['Toma_#'].tolist()

    for codigo_toma in codigos_tomas:
        dia_actual.append(codigo_toma)

        if len(dia_actual) == max_tomas_por_dia:
            agenda_dias.append(list(dia_actual))
            dia_actual = []

    # Incorpora las tomas restantes que no completaron una jornada completa
    if dia_actual:
        agenda_dias.append(list(dia_actual))

    return agenda_dias


# --- Paso 2: Función de Impresión Reutilizable ---
def imprime_agenda(agenda, df_clean):
    """
    Despliega de manera detallada y perfectamente alineada la planificación,
    indicando las tomas y los actores requeridos de forma visualmente limpia.
    """
    columnas_actores = [col for col in df_clean.columns if col.startswith('Actor_')]

    for num_dia, tomas_dia in enumerate(agenda, start=1):
        print(f"\n🔹 Día {num_dia}: {len(tomas_dia)} tomas asignadas -> {tomas_dia}")
        print("   " + "─" * 70)

        for toma in tomas_dia:
            try:
                fila = df_clean[df_clean['Toma_#'] == toma].iloc[0]
                actores_activos = [col for col in columnas_actores if col in df_clean.columns and fila[col] == 1]
            except Exception as error:
                actores_activos = [f"Error: {error}"]

            # Convertimos la lista de actores a un texto plano separado por comas
            actores_str = ", ".join(actores_activos)

            # Usamos un ancho fijo (ej. 12 espacios para la toma) para que todo caiga en columna vertical exacta
            print(f"    └─ {toma:<10} ➔  [{actores_str}]")

        print("   " + "─" * 70)

# Ejecucion --- >

# 1. Crea agenda inicial
agenda_inicial = crear_agenda_doblaje(df_clean, max_tomas_por_dia=6)
# 2. imprime Agenda
print("--- IMPRIMIENDO AGENDA  ---")
imprime_agenda(agenda_inicial, df_clean)
````

# **Paso 5: El Algoritmo de optimización**
   - Costo de la funcion
   - Algoritmo de simulacion(Simulating Anaeling)


````python
# Definición de la Función de Coste
def calcular_costo_agenda(agenda_dias, matriz_actores):
    """
    Calcula el coste total de la planificación evaluando las apariciones
    necesarias de los actores a lo largo de las jornadas de grabación,
    utilizando los códigos de las tomas.
    """
    costo_total = 0

    for tomas_dia in agenda_dias:
        # Salta al próximo ciclo en caso de que el día esté vacío
        if not tomas_dia:
            continue

        # Selecciona las filas del DataFrame donde el código de la toma
        # coincida con los que están asignados en la agenda del día actual
        df_dia = matriz_actores[matriz_actores['Toma_#'].isin(tomas_dia)]

        # Determina si el actor participa al menos en una toma durante la jornada
        # (Omitimos la columna 'Toma_#' con .iloc[:, 1:] para sumar únicamente los actores)
        presencia_actor_dia = (df_dia.iloc[:, 1:].sum() > 0)

        # Suma cuántos actores distintos asistieron en este día y acumula el coste total
        costo_total += presencia_actor_dia.sum()

    return costo_total

# --- Prueba del Coste Base ---
# 1. Creación de la agenda inicial con los códigos de tomas
#initial_schedule = crear_agenda_doblaje(df_clean, max_tomas_por_dia=6)

# 2. Cálculo del coste base para el horario sin optimizar
#baseline_cost = calcular_costo_agenda(initial_schedule, df_clean)
#print(f"\nCosto base (sin optimizar): {baseline_cost}")
````

````python
# --- Paso 6: Algoritmo de Optimización por Recocido Simulado ---
def simulated_annealing_schedule(initial_schedule, df, initial_temp=100.0, cooling_rate=0.95, max_iterations=1000):
    """
    Optimiza la distribución de las tomas en la agenda aplicando
    la técnica de optimización estocástica de Recocido Simulado.
    """
    current_schedule = copy.deepcopy(initial_schedule)
    current_cost = calcular_costo_agenda(current_schedule, df)

    best_schedule = copy.deepcopy(current_schedule)
    best_cost = current_cost

    temperature = initial_temp

    for iteration in range(max_iterations):
        # 1. Perturbación: Genera un estado vecino mediante un intercambio aleatorio de tomas entre diferentes días
        neighbor_schedule = copy.deepcopy(current_schedule)

        day_indices = [i for i, day in enumerate(neighbor_schedule) if len(day) > 0]
        if len(day_indices) < 2:
            continue

        d1, d2 = random.sample(day_indices, 2)
        if neighbor_schedule[d1] and neighbor_schedule[d2]:
            t1_idx = random.randint(0, len(neighbor_schedule[d1]) - 1)
            t2_idx = random.randint(0, len(neighbor_schedule[d2]) - 1)

            # Intercambio de tomas entre jornadas
            neighbor_schedule[d1][t1_idx], neighbor_schedule[d2][t2_idx] = \
                neighbor_schedule[d2][t2_idx], neighbor_schedule[d1][t1_idx]

        # 2. Evaluación del coste de la solución vecina
        neighbor_cost = calcular_costo_agenda(neighbor_schedule, df)
        cost_diff = neighbor_cost - current_cost

        # 3. Criterio de aceptación basado en la mejora o en la probabilidad de temperatura
        if cost_diff < 0 or random.random() < math.exp(-cost_diff / temperature):
            current_schedule = neighbor_schedule
            current_cost = neighbor_cost

            # Registro de la mejor solución encontrada hasta el momento
            if current_cost < best_cost:
                best_schedule = copy.deepcopy(current_schedule)
                best_cost = current_cost

        # 4. Reducción progresiva de la temperatura
        temperature *= cooling_rate

    return best_schedule, best_cost
````


# **Pruebas con Diferentes Juegos de Datos**
Para validar la robustez del modelo, se han generado y probado diferentes escenarios de entrada variando la cantidad de tomas y la densidad de participación de los actores.











1. **Escenario Base:** 30 tomas y 10 actores (matriz leida de una archivo .xls).
