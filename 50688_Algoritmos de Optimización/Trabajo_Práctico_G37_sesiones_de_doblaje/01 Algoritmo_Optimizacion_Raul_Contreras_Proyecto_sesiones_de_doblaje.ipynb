### 1 Análisis de Complejidad
La complejidad temporal de este algoritmo de Recocido Simulado depende principalmente del número de iteraciones ($I$), así como de la cantidad de tomas ($T$) y actores ($A$). En cada iteración se evalúa el coste recorriendo la matriz completa, por lo que el tiempo total de ejecución resulta de multiplicar las iteraciones por el tamaño de los datos, demostrando ser un algoritmo rápido y eficiente para el volumen dado.
- Complejidad temporal: Está determinada por el número de iteraciones ($I$) destinadas a mejorar la agenda, la cantidad de tomas ($T$) y el total de actores ($A$).
- Cálculo de coste: En cada evaluación de una agenda, el algoritmo recorre la totalidad de las tomas y actores presentes en la matriz.
- Bucle principal: Dado que este cálculo de coste se ejecuta en cada una de las 1,000 iteraciones, el tiempo global resulta de multiplicar las iteraciones por el tamaño de la matriz.
- Conclusión: Se trata de un algoritmo rápido y eficiente para el volumen de datos manejado en este proyecto de doblaje.

# **Paso 4: El Creamos la agenda y la imprimimos**
   - Crear Agenda
   - Imprimir Agenda

````python
# Función para Crear la Agenda  
def crear_agenda_doblaje(matriz_actores, max_tomas_por_dia=6):
    """
    Agrupa las tomas disponibles en jornadas diarias utilizando los códigos
    reales de la columna 'Toma_#', respetando el límite máximo establecido por día.
    """
    agenda_dias = []
    dia_actual = []

    # Extrae directamente los códigos de texto de la columna 'Toma_#'
    codigos_tomas = matriz_actores['Toma_#'].tolist()

    for codigo_toma in codigos_tomas:
        dia_actual.append(codigo_toma)

        if len(dia_actual) == max_tomas_por_dia:
            agenda_dias.append(list(dia_actual))
            dia_actual = []

    # Incorpora las tomas restantes que no completaron una jornada completa
    if dia_actual:
        agenda_dias.append(list(dia_actual))

    return agenda_dias


# Función de Impresión 
def imprime_agenda(agenda, df_clean):
    """
    Despliega de manera detallada y perfectamente alineada la planificación,
    indicando las tomas y los actores requeridos de forma visualmente limpia.
    """
    columnas_actores = [col for col in df_clean.columns if col.startswith('Actor_')]

    for num_dia, tomas_dia in enumerate(agenda, start=1):
        print(f"\n🔹 Día {num_dia}: {len(tomas_dia)} tomas asignadas -> {tomas_dia}")
        print("   " + "─" * 70)

        for toma in tomas_dia:
            try:
                fila = df_clean[df_clean['Toma_#'] == toma].iloc[0]
                actores_activos = [col for col in columnas_actores if col in df_clean.columns and fila[col] == 1]
            except Exception as error:
                actores_activos = [f"Error: {error}"]

            # Convertimos la lista de actores a un texto plano separado por comas
            actores_str = ", ".join(actores_activos)

            # ancho fijo ej. 12 espacios para la toma
            print(f"    └─ {toma:<10} ➔  [{actores_str}]")

        print("   " + "─" * 70)

# Ejecucion --- >

# 1. Crea agenda inicial
agenda_inicial = crear_agenda_doblaje(df_clean, max_tomas_por_dia=6)
# 2. imprime Agenda
print("--- IMPRIMIENDO AGENDA  ---")
imprime_agenda(agenda_inicial, df_clean)
````

 # **Paso 5: Calcula El costo de la Optimizacion de la Agenda**
   - Costo 

````python
# Definición de la Función de Coste
def calcular_costo_agenda(agenda_dias, matriz_actores):
    """
    Calcula el coste total de la planificación evaluando las apariciones
    necesarias de los actores a lo largo de las jornadas de grabación,
    utilizando los códigos de las tomas.
    """
    costo_total = 0

    for tomas_dia in agenda_dias:
        # Salta al próximo ciclo en caso de que el día esté vacío
        if not tomas_dia:
            continue

        # Selecciona las filas del DataFrame donde el código de la toma coincida con los que están asignados en la agenda del día actual
        df_dia = matriz_actores[matriz_actores['Toma_#'].isin(tomas_dia)]

        # Determina si el actor participa al menos en una toma durante la jornada se omite columna'Toma_#' con .iloc[:, 1:] 
        # para sumar únicamente los actores)
        presencia_actor_dia = (df_dia.iloc[:, 1:].sum() > 0)

        # Suma cuántos actores distintos asistieron en este día y acumula el coste total
        costo_total += presencia_actor_dia.sum()

    return costo_total
````

# ** Paso 6: Algoritmo Fuerza bruta**
Un algoritmo de fuerza bruta genera todas las combinaciones y permutaciones posibles de asignar las tomas a los diferentes días, evaluando la función de coste en cada una de ellas para quedarse con la que tenga el menor coste.

- Prueba todas las combinaciones posibles sin importar lo ineficiente que sea. Explora el espacio de búsqueda completo para garantizar que encuentra la solución óptima, es de complejidad alta (exponencial),  se vuelve inviable con problemas grandes.

````python
#
# Permutaciones_plana:
# Función que recibe una lista simple de elementos  ['Toma_01',...] y devolver una lista con todas las combinaciones posibles
# de esos elementos. No sabe nada de días, Tomas, solo de lista de tomas
#
def generar_todas_las_permutaciones_plana(tomas):
    if len(tomas) <= 1:
        return [tomas]
    permutaciones_totales = []
    for i in range(len(tomas)):
        elemento_actual = tomas[i]
        elementos_restantes = tomas[:i] + tomas[i+1:]
        for perm in generar_todas_las_permutaciones_plana(elementos_restantes):
            permutaciones_totales.append([elemento_actual] + list(perm))
    return permutaciones_totales

    
#
# Permutaciones Agendas(agenda_original) 
# Esta función "llama" a permutaciones_plana y gestiona el problema del doblaje.
# Recibe la agenda compleja una lista de listas: [['T1', 'T2'], ['T3', 'T4']].
# Llama a la función plana para que permute las tomas como si fueran una sola lista larga (['T1', 'T2', 'T3', 'T4']).
# Vuelve a reconstruir la estructura: Una vez que la función plana devuelve todas las listas de tomas permutadas
# ej: ['T3', 'T1', 'T4', 'T2']), con esta lista larga la vuelve a cortar en secciones para devolverla al formato original [['T3', 'T1'], ['T4', 'T2']].
def generar_permutaciones_agendas(agenda_original):
    print("-> Aplanando tomas de la agenda...")
    tomas_planas = [toma for dia in agenda_original for toma in dia]
    print(f"-> Total de tomas a permutar: {len(tomas_planas)}")
    
    print("-> Generando permutaciones (esto puede tardar según el tamaño)...")
    permutaciones_tomas = generar_todas_las_permutaciones_plana(tomas_planas)
    print(f"-> ¡Permutaciones generadas! Total encontradas: {len(permutaciones_tomas)}")
    
    agendas_permutadas = []
    for perm in permutaciones_tomas:
        nueva_agenda = []
        indice = 0
        for dia in agenda_original:
            longitud_dia = len(dia)
            nuevo_dia = list(perm[indice:indice + longitud_dia])
            nueva_agenda.append(nuevo_dia)
            indice += longitud_dia
        agendas_permutadas.append(nueva_agenda)
        
    return agendas_permutadas


# Algoritmo principal de Fuerza Bruta por días
def fuerza_bruta_agenda_por_dias(agenda_base, df):
    mejor_agenda = None
    menor_coste = float('inf')
    
    todas_las_opciones = generar_permutaciones_agendas(agenda_base)
    total_opciones = len(todas_las_opciones)
    print(f"-> Iniciando evaluación de {total_opciones} agendas candidatas...")

    for i, agenda_candidata in enumerate(todas_las_opciones):
        # Muestra un aviso cada 5,000 iteraciones para que veas que avanza
        if i > 0 and i % 5000 == 0:
            print(f"   > Evaluadas {i} de {total_opciones} agendas...")
            
        coste = calcular_costo_agenda(agenda_candidata, df)
        if coste < menor_coste:
            menor_coste = coste
            mejor_agenda = agenda_candidata

    print("-> ¡Evaluación completa!")
    return mejor_agenda
    

# Ejecución controlada (Muestra acotada para evitar saturación factorial O(n!))
agenda_inicial = crear_agenda_doblaje(df_clean, max_tomas_por_dia=3)
agenda_muestra = agenda_inicial[:2]  # Restricción a 2 días / ~6 tomas (720 permutaciones) para ejecución óptima


print("Iniciando búsqueda por Fuerza Bruta con vuestra agenda...")
mejor_agenda_fb = fuerza_bruta_agenda_por_dias(agenda_muestra, df_clean)
coste_fb = calcular_costo_agenda(mejor_agenda_fb, df_clean)

# Resultados de la ejecución
print("\n Resultados de Fuerza Bruta --->")
print(f"Mejor Agenda Encontrada: {mejor_agenda_fb}")
print(f"Coste Optimizado: {coste_fb}")
   
````

# **Paso 7: El Algoritmo  Recocido Simulado(Simulating Anaeling) **
   - Costo de la funcion
   - Algoritmo de simulacion(Simulating Anaeling)


````python
# --- Paso 6: Algoritmo de Optimización por Recocido Simulado ---
def simulated_annealing_schedule(initial_schedule, df, initial_temp=100.0, cooling_rate=0.95, max_iterations=1000):
    """
    Optimiza la distribución de las tomas en la agenda aplicando
    la técnica de optimización estocástica de Recocido Simulado.
    """
    current_schedule = copy.deepcopy(initial_schedule)
    current_cost = calcular_costo_agenda(current_schedule, df)

    best_schedule = copy.deepcopy(current_schedule)
    best_cost = current_cost

    temperature = initial_temp

    for iteration in range(max_iterations):
        # 1. Perturbación: Genera un estado vecino mediante un intercambio aleatorio de tomas entre diferentes días
        neighbor_schedule = copy.deepcopy(current_schedule)

        day_indices = [i for i, day in enumerate(neighbor_schedule) if len(day) > 0]
        if len(day_indices) < 2:
            continue

        d1, d2 = random.sample(day_indices, 2)
        if neighbor_schedule[d1] and neighbor_schedule[d2]:
            t1_idx = random.randint(0, len(neighbor_schedule[d1]) - 1)
            t2_idx = random.randint(0, len(neighbor_schedule[d2]) - 1)

            # Intercambio de tomas entre jornadas
            neighbor_schedule[d1][t1_idx], neighbor_schedule[d2][t2_idx] = \
                neighbor_schedule[d2][t2_idx], neighbor_schedule[d1][t1_idx]

        # 2. Evaluación del coste de la solución vecina
        neighbor_cost = calcular_costo_agenda(neighbor_schedule, df)
        cost_diff = neighbor_cost - current_cost

        # 3. Criterio de aceptación basado en la mejora o en la probabilidad de temperatura
        if cost_diff < 0 or random.random() < math.exp(-cost_diff / temperature):
            current_schedule = neighbor_schedule
            current_cost = neighbor_cost

            # Registro de la mejor solución encontrada hasta el momento
            if current_cost < best_cost:
                best_schedule = copy.deepcopy(current_schedule)
                best_cost = current_cost

        # 4. Reducción progresiva de la temperatura
        temperature *= cooling_rate

    return best_schedule, best_cost
````
